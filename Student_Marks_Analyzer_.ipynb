```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

subjects = np.array([
    "Python",
    "Java",
    "DBMS",
    "Maths",
    "Web Development"
])

marks = np.random.randint(0, 101, size=(50, 5))

print("Marks of 50 Students:")
print(marks)
```

```python
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total Marks Entries:", marks.size)
print("Data Type:", marks.dtype)
```

```python
# First 10 students
print(marks[:10])

# First student's marks
print(marks[0])

# First student's Python marks
print(marks[0, 0])

# Last subject ke marks for all students
print(marks[:, -1])

# First 5 students and first 3 subjects
print(marks[:5, :3])
```

```python
total_marks = np.sum(marks, axis=1)
average_marks = np.mean(marks, axis=1)

for i in range(50):
    print(
        f"Student {i+1}: "
        f"Total = {total_marks[i]}, "
        f"Average = {average_marks[i]:.2f}"
    )
```

```python
topper_index = np.argmax(total_marks)

print("Overall Topper: Student", topper_index + 1)
print("Total Marks:", total_marks[topper_index])
print("Average:", average_marks[topper_index])
```

```python
subject_averages = np.mean(marks, axis=0)

for i in range(len(subjects)):
    print(subjects[i], "Average:", round(subject_averages[i], 2))
```

```python
for i in range(len(subjects)):
    topper_index = np.argmax(marks[:, i])

    print(
        subjects[i],
        "- Topper: Student",
        topper_index + 1,
        "- Marks:",
        marks[topper_index, i]
    )
```

```python
failed_mask = np.any(marks < 40, axis=1)

failed_students = np.where(failed_mask)[0] + 1

print("Students with at least one subject below 40:")
print(failed_students)
```

```python
high_scorers_mask = marks > 90

students, subject_indices = np.where(high_scorers_mask)

for student, subject in zip(students, subject_indices):
    print(
        f"Student {student+1} scored "
        f"{marks[student, subject]} in {subjects[subject]}"
    )
```

```python
grades = np.select(
    [
        average_marks >= 90,
        average_marks >= 75,
        average_marks >= 60,
        average_marks >= 40
    ],
    ["A", "B", "C", "D"],
    default="F"
)

for i in range(50):
    print(
        f"Student {i+1}: "
        f"Average = {average_marks[i]:.2f}, "
        f"Grade = {grades[i]}"
    )
```

```python
passed_students = np.all(marks >= 40, axis=1)

pass_count = np.sum(passed_students)
pass_percentage = (pass_count / len(marks)) * 100

print("Total Students:", len(marks))
print("Passed Students:", pass_count)
print("Failed Students:", len(marks) - pass_count)
print("Pass Percentage:", round(pass_percentage, 2), "%")
```

```python
plt.figure(figsize=(10, 5))

plt.bar(subjects, subject_averages)

plt.title("Subject-Wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
